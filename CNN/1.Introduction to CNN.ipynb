### **What is CNN?**

Convolutional Neural Network (CNN) is a Deep Learning architecture mainly used for:

- Image Classification
- Object Detection
- Face Recognition
- Medical Imaging
- OCR
- Video Processing

CNN automatically learns:

- edges
- textures
- shapes
- patterns
- objects

from images.

### **Why CNN is Important**

Images contain large amounts of pixel data.

For example:

```python
224 × 224 × 3 = 150,528 pixels
```

A traditional neural network connecting every pixel to neurons would require millions of parameters.

CNN solves this problem through:

- Local connectivity
- Shared weights
- Automatic feature extraction
- Dimensionality reduction

This makes CNN computationally efficient and highly accurate for image-related tasks.

### **Problems with Traditional Neural Networks**

Traditional Artificial Neural Networks (ANNs) face several limitations when handling image data.

Major Problems
1. Huge Number of Parameters

For high-resolution images:

1000 × 1000 × 3 = 3,000,000 inputs

This leads to enormous computational complexity.

### **Basic CNN Architecture**

Input Image → Convolution → ReLU → Pooling → Flatten → Dense Layer → Output

**Important Layers**

**(a) Convolution Layer**

Applies filters/kernels to detect features.

Example:

Edges
Shapes
Textures

Formula: [Output = (Input * Kernel) + Bias]


**Important Parameters:**

Number of filters
- Kernel size
- Stride
- Padding
